# Function

```python
# Def: A function is a block of reusable code that perform a specific task.

def fun():   # definig the function
    print("Welcome to functions")

    print("Hello world")

fun()  # calling the function until and unless we dont calll the function it will not extecute
fun()  # u call the function as many times as u want
```

```python
# ex-2 : passing one parameter in the function

def fun(val):   # definig the function and passing value to a function
    print("Welcome to functions", val)

    print("Hello world")

fun(100)  # calling the function
```

```python
# ex-3 : passing parameter in the function

def fun(val1, val2):   # definig the function and passing value to a function
    print("Welcome to functions", val1)

    print("Hello world", val2)

fun(100,200)  # calling the function
```

```python
# ex-4 : nested function : function inside the function (we can only do in python not in other languages)

def outer():
    print("Iam outer function")
    def inner():
        print("Iam inner function")
    inner()
outer()
```

```python
def bulding():
    print("we are here")
    def room1():
        print("pavan: iam here")
    room1()  # calling from inside the bulding function
bulding()
```

```python
# ex-5 : Return function

def fun(a,b):
    c = a + b
    return c    # return single value
    
val = fun(10,20)
print(val)
```

```python
def fun(a,b):
    c = a + b
    d = a - b
    e = a * b
    return c,d,e   # returing multiple values
    
val = fun(20,10)
print(val)
```

# Default Parameters

```python
# assign a default value to a "var" variable
def fun(var = "Nothing"):
    print(var)

fun()
fun(100)  # here we are passing value
```

```python
def fun2(x = 10, y = 20):
    print(x + y)

fun2()
fun2(100)
fun2(100, 200)
```

# passing arbitary number of arguments using *args

```python
def fun(*args):
    for i in args:
        print(i)

fun()
fun("chinki")
fun("chinki","Minki") # u can ass any numbers of arguments to the function by using *args
```

```python
def fun(*var):
    for i in var:
        print(i)

fun()
fun("chinki")
fun("chinki","Minki")
```

# Lambda function

```python
# DEF: Lambda function a single line of code that is used to perform a short task without defining a full function with def.

# A lambda function don't have any name
# Although lambda function can receive arguments and process the data
# lambda function are popularly used inside the other function to perform small task
```

```python
# we access it my obj and we use "lambda" as keyword

obj = lambda a : a + 100

print(obj(10))
```

```python
obj = lambda a,b : a + b

print(obj(100,200))
```

```python
obj1 = lambda a,b,c : a * b * c

print(obj1(10,2,5))
```

```python
# lambda function are popularly used inside the other function to perform small task

def fun(p):
    return lambda a : a**p

obj1 = fun(10)
print(obj1(2))
print(obj1(3))
```

```python
# using lambda function inside the other function
def fun(p):
    obj = lambda a : a**p
    return obj

obj = fun(10)

print(obj(2))
print(obj(3))
```

```python
def fun(p):
    obj = lambda a : a**p if a%2==0 else a+p
    return obj

obj = fun(2)

print(obj(2))
print(obj(3))
```

```python
# sort the bellow given list of tuple

students = [("Pavan",80),("Dinesh",75),("Ak",78)]

students.sort()
print(students)
```

```python
# sort 
# lambda func popularly used with sorted() function

students = [("Pavan",80),("Dinesh",75),("Ak",78)]

sorted_data = sorted(students, key = lambda a : a[1])
print(sorted_data)
```

```python
students = [("Pavan",80,5),("Dinesh",75,15),("Ak",78,10)]

sorted_data = sorted(students, key = lambda a : a[2])   # sort based a index value 
print(sorted_data)
```

```python
students = {"Pavan":80,"Dinesh":75,"Ak":78}
data = students.items() # items gives all the data as list of tuple
print(data)

sorted_data = sorted(data, key = lambda a : a[1])   # sort based a index value 
print(sorted_data)
```
