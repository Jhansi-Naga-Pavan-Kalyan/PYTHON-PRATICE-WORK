```python
#EX-1: struture Multilevel inhheritance

class greatGF:
    pass
    
class grandfather(greatGF):
    pass
    
class father(grandfather):
    pass
    
class son(father):
    pass
```

```python
class greatGF:
    def Gold(self):
        print("1000 KG")
    
class grandfather(greatGF):
    def Land(self):
        print("500 Acr")
    
class father(grandfather):
    def Car(self):
        print("Lamborgini")
    
class son(father):
    def Bike(self):
        print("RE bike")

obj_son = son()
obj_son.Bike()
obj_son.Car()
obj_son.Land()
obj_son.Gold()
```

```python
#Ex-3 : super() function

class greatGF:
    def Gold(self):
        print("1000 KG")
    
class grandfather(greatGF):
    def Land(self):
        print("500 Acr")
    
class father(grandfather):
    def Car(self):
        print("Lamborgini")
    
class son(father):
    def Bike(self):
        print("RE bike")
        super().Car()
        super().Land()
        super().Gold()

obj_son = son()
obj_son.Bike()
```

# 4. Hierarichal inheritance

```python
# Many child one parent

#Ex-1: Structure of hierarichal: one parent two child
class superhero:
    pass
    
class sinchain(superhero):
    pass

class Doreman(superhero):
    pass
```

```python
class superhero:
    def intro(self):
        print("Superhero powers")
    
class sinchain(superhero):
    def comedy(self):
        print("Himavari")

class Doreman(superhero):
    def pocket(self):
        print("Nobita Friend")

obj_dore = Doreman()
obj_dore.intro()  # doreman can't acess the properties of sinchain
```

```python
class superhero:
    def __init__(self):
        print("intro CF")
    def intro(self):
        print("Superhero powers")
    
class sinchain(superhero):
    def __init__(self):
        print("sinchain CF")
    def comedy(self):
        print("Himavari")

class Doreman(superhero):
    
    def __init__(self):
        print("doreman CF")
    def pocket(self):
        print("Nobita Friend")
        
obj_dore = Doreman()
```

```python
class superhero:
    def __init__(self):
        print("intro CF")
    def intro(self):
        print("Superhero powers")
    
class sinchain(superhero):
    def __init__(self):
        print("sinchain CF")
    def comedy(self):
        print("Himavari")

class Doreman(superhero):
    
    def __init__(self):
        print("doreman CF")
    def pocket(self):
        print("Nobita Friend")
        super().intro()
        
obj_dore = Doreman()
obj_dore.pocket()

#method resolution order
print([i.__name__ for i in Doreman.__mro__])
```

# 5. Hybrid inheritance

```python
# Struture of hybrid inheritance
class A:
    pass
    
class B(A):
    pass

class C(B):
    pass

class D(B):
    pass

class E(C,D):
    pass
```

```python
class A:
    A_var = "A-class"
    
class B(A):
    B_var = "B-class"

class C(B):
    C_var = "C-class"

class D(B):
    D_var = "D-class"

class E(C,D):
    E_var = "E-class"

obj_E = E()
print(obj_E.E_var)
```

```python
class A:
    A_var = "A-class"
    def __init__(self):
        print("A CF")
    
class B(A):
    B_var = "B-class"
    def __init__(self):
        print("B CF")

class C(B):
    C_var = "C-class"
    def __init__(self):
        print("C CF")

class D(B):
    D_var = "D-class"
    def __init__(self):
        print("D CF")

class E(C,D):
    E_var = "E-class"
    def __init__(self):
        print("E CF")

obj_E = E()
```

```python
# super()

class A:
    A_var = "A-class"

    def A_output(self):
        print("this is A output")
    
class B(A):
    B_var = "B-class"

    def B_output(self):
        print("this is B output")

class C(B):
    C_var = "C-class"

    def C_output(self):
        print("this is C output")

class D(B):
    D_var = "D-class"

    def D_output(self):
        print("this is D output")
        super().A_output()
    

class E(C,D):
    E_var = "E-class"

    def E_output(self):
        print("this is E output")
        super().A_output()

obj_E = E()
obj_E.E_output()
```

```python
# super()

class A:
    A_var = "A-class"

    def A_output(self):
        print("this is A output")
    
class B(A):
    B_var = "B-class"

    def A_output(self):
        print("this is B output")
        super().A_output()

class C(B):
    C_var = "C-class"

    def A_output(self):
        print("this is C output")
        super().A_output()
        

class D(B):
    D_var = "D-class"

    def A_output(self):
        print("this is D output")
        super().A_output()

class E(C,D):
    E_var = "E-class"

    def A_output(self):
        print("this is E output")
        super().A_output()

obj_E = E()
obj_E.A_output()

print([i.__name__ for i in E.__mro__])
```
