# Encapsulation

```python
#Putting data and function together inside a class/single unit 
# only the function inside the class can access the data
```

```python
# DEF: Encapulation refers to the binding the data(attributes) and functions into one unit(class)
# Encapulation restrict the direct access of the data.
```

```python
class Bike:
    No_tyre = 2    # class variable / Attribute / data

    def __init__(self, name, price):
        self.bike_name = name     # instance variable
        self.bike_price = price
        print(self.bike_name,self.bike_price)

    def display(self):
        print(self.No_tyre, self.bike_name, self.bike_price)

obj1 = Bike("KTM",500000)
```

```python
# we can access the variables/data/attribute through object outside the class 
# we can access the variables/data/attribute through function defined inside the class 

class Bike:
    No_tyre = 2    # class variable / Attribute / data

    def __init__(self, name, price):
        self.bike_name = name     # instance variable
        self.bike_price = price

    def display(self):
        print(self.No_tyre, self.bike_name, self.bike_price)     # Access data through inside the function

obj1 = Bike("KTM",500000)

obj1.display()          # Access data through object outside the class

print(obj1.No_tyre)
print(obj1.bike_name)
print(obj1.bike_price)
```

# Access Specifers / types of attributes in encapsulation

```python
# 1. Public : these variables can be accesssed inside/outside/child the class
             # Public variables don't start with underscore
# 2. Protected : these variables can be accessed only inside/child class
                 # protected variable NOT INTENDED to access outside the class
                # protected variables start with SINGLE underscore
# 3. Private : these variables can be accessed only inside class
               # NOT INTENDED to access outside/child class
               # Private variables start with DOUBLE underscore
```

```python
# All variables can be accesssed inside the class

class marvels:
    def __init__(self, name, power, weapon):
        self.Name = name          # Public instant variable
        self._Power = power       # Protected instant variable
        self.__Ayudham = weapon   # Private instant variable

    def show(self):
        print(self.Name,self._Power,self.__Ayudham)

obj1 = marvels("Thor", 1000, "Hammer")
obj1.show()
```

```python
class marvels:
    def __init__(self, name, power, weapon):
        self.Name = name          # Public instant variable
        self._Power = power       # Protected instant variable
        self.__Ayudham = weapon   # Private instant variable

    def show(self):
        print(self.Name,self._Power,self.__Ayudham)

obj1 = marvels("Thor", 1000, "Hammer")
obj1.show()

print(obj1.Name)
print(obj1._Power)   # although Protected instant can access outside the class but we should'nt access it
# IN python public, protected, private concept is used just for Sign/Warning
print(obj1._marvels__Ayudham)  # obj.classname__privatevariable
```
