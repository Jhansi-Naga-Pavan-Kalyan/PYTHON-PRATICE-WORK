# 1. Single Inheritance : One parent one child

```python
## Ex-1 : structure of single inheritance
# parent
class hyd_food:
    taste = "very very very Spicy"     # property/class varaible

    def Biryani(self):
        print("very very tasty Biryani")

#child class
class PistaHouse(hyd_food):
    pass

obj_child = PistaHouse()  # it is the object of child class

# child class object can access the varaible/properties & function of parent class

print(obj_child.taste)  # calling parent class varaible
obj_child.Biryani()    # calling parent class function with object of child
```

```python
# super() is used to access the parent class's methods or constructor from the child class.
```

```python
#Ex-2 : super() function
# super() : used to access the parent properties and function from the inside the child class

# parent
class hyd_food:
    taste = "very very very Spicy"     # property/class varaible
 
    def Biryani(self):      # parent class function
        print("very very tasty Biryani")

#child class
class PistaHouse(hyd_food):
    def display(self):
        super().Biryani()
        print(super().taste)    # accessing parent property from child class

obj_child = PistaHouse()  # it is the object of child class
obj_child.display()     # accessing parent function from child class
```

```python
# super() with __init__() : It is also commonly used to call the parent constructor
```

```python
# Ex-3 : constructor function
# parent
class hyd_food:
    taste = "very very very Spicy"     # property/class varaible

    def Biryani(self):
        print("very very tasty Biryani")

#child class
class PistaHouse(hyd_food):
    def __init__(self):
        print("I am PistaHouse class")

obj_child = PistaHouse()    # it is the object of child class
```

```python
# if child class contructor function & parent class contructor function their it exectues child class
class hyd_food:
    taste = "very very very Spicy"# property/class varaible

    def __init__(self):
        print("father")     # parent class contructor function

    def Biryani(self):
        print("very very tasty Biryani") 

#child class
class PistaHouse(hyd_food):
    
    def __init__(self):
        print("child")   

obj_child = PistaHouse()
```

# 2. Multiple inheritance : Many parents one Child

```python
# Example
class hyd_food:
    taste = "very very very Spicy"# property/class varaible

    def __init__(self):
        print("I love biryani")     # parent class contructor function
        super().__init__()

    def Biryani(self):
        print("very very tasty Biryani") 

class pune_food:
    taste = "very very very yummy"

    def __init__(self):
        print("I love vada pav")

#child class
class PistaHouse(hyd_food,pune_food):
    def __init__(self):
        print("I like biryani & vada pav")
        super().__init__()

obj_child = PistaHouse()
obj_child.Biryani()
obj_child.taste
obj_child1 = hyd_food()
obj_child1 = obj_child.taste
```

```python
# EX-1 : design the structure

class A:
    def display(self):
        print("I Am A class")

class B:
    def show(self):
        print("I Am B class")

class C(A,B):
    pass

obj_C = C()
obj_C.display()
obj_C.show()
```

```python
# Ex-4: super() function
class A:
    def display(self):
        print("I Am A class")

class B:
    def show(self):
        print("I Am B class")

class C(A,B):

    def hello(self):
        super().display()
        super().show()

obj_C = C()
obj_C.hello()
```

```python
# Ex-4.2: super() function
class A:
    def display(self):
        print("I Am A class")

class B:
    def display(self):
        print("I Am B class")

class C(A,B):

    def hello(self):
        super().display()
        super().display()

obj_C = C()
obj_C.hello()
```

```python
# Ex-4.3: super() function : follow the  MRO approach (method resolution order)
class A:
    def display(self):
        print("I Am A class")
        super().display()

class B:
    def display(self):
        print("I Am B class")

class C(A,B):

    def hello(self):
        super().display()


obj_C = C()
obj_C.hello()
```
