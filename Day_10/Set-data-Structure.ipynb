# creating different types of set

```python
empty_set = set()

# set function is used to convert list,tuple of elements to set
set2 = set([10,20,30])
set3 = set((10,20,30))

#frozen sets are immutable (cannot perform insert, delete, update)
fro_set = frozenset([10,20,30,40])

nested_set = {(1,23,4),"Hello"} # tuple can be store in the set
#nested_set1 = {[1,23,4]1,2,3} # list inside set cannot be store 
#nested_set2 = {{1,23,4}1,2,3} # set inside set cannot be store
nested_set3 = {(1,23,4),"Hello",frozenset([10,20,30,40])} # frozenset can be store in the set

# IMP: Mutable items cannot be store inside the store
# Only immutable items can be store in the set

print(empty_set)
print(set2)
print(set3)
print(fro_set)
print(s1)
print(nested_set)
print(nested_set3)
```

# insert in set (add(), update())

```python
set1 = {10, 20.5, "Hello"}

# add() function is used to insert one item in the set
set1.add("world")

print(set1)
```

```python
set1.add((10,20,30,40)) # as it  ts a tuple its a single item only

print(set1)
```

```python
# update is used for inserting multiple items together

set1.update((1000,2000))

print(set1)
```

```python
set1.update({(1000,2000)})

print(set1)
```

```python
# multiple items can be passed to update() function in a set or tuple or list
# update function is not storing list inside the set its taking elements from the list

s1 = {101,102,103}
t1 = (201,202,203)
l1 = [301,302,303]

set1.update(s1)
set1.add(t1)
print(set1)
```

# delete operation on set (remove(), discard(), pop(), clear(), del)

```python
dinner_set = {"Bagara rice", "Biryani", "papu", "Chapati", "fish", "Mutton", "prawns"}

# remove() functionn delete the one item whose name is provided
# remove() will raise an ERROR if item don't exist

dinner_set.remove("prawns")     #if item is not their it will show the error

print(dinner_set)
```

```python
# discard() functionn delete the one item whose name is provided
# discard() DO NOT raise an ERROR if item don't exist

dinner_set.discard("Biryani")   #if item is not their it will not show the error


print(dinner_set)
```

```python
# pop() will delete any random item
dinner_set.pop()

print(dinner_set)
```

```python
# clear() will delete all the items in the set
dinner_set.clear()

print(dinner_set)
```

```python
del dinner_set # it will delete the whole set
```

# apply popular inbuilt functions on set

```python
set2 = {10, 20, 30, 2, 3, 5 ,1000, 4}

print(len(set2))
print(min(set2))
print(max(set2))
print(sum(set2))
```

```python
l1 = sorted(set2)    # we use sorted() nd sorted function gives list
print(l1)
```

```python
l1 = sorted(set2, reverse = True)
print(l1)
```

# containment operation on set

```python
south_festivals = {"Sankranti", "Bhogi", "Ugadi", "Onam", "Dussahra"}

if "Holi" in south_festivals:
    print("Happy Holi")
elif "Sankranti" not in south_festivals:
    print("Happy Sankranti")
elif "Ugadi" in south_festivals:
    print("Happy Ugadi")
```
