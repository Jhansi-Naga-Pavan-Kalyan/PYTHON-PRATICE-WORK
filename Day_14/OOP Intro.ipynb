```python
# EX-1: Structure of Class

#class is a keyword, foood is the class name
# class variable: the value of class variable will remains SAME for all the objects
# class variable value cannot be change by object but it can only change by using class name
class food:    
    food_name = "Biryani"   # class variable we defined

obj = food()    # creating the object (obj) of food() class
obj.food_name  # accessing the class variable
```

```python
# we can create multiple objects
class food:    
    food_name = "Biryani"   # class variable we defined

obj1 = food()    # creating the object (obj) of food() class
print(obj1.food_name)  # accessing the class variable

obj2 = food()
print(obj2.food_name) # class object (obj2) can access the class data

# we cannot access class variable diectly we have to access by using object
```

```python
# class variable value cannot be change by object
# class variable CAN be only change by class name
class food:    
    food_name = "Biryani"   # class variable we defined

obj1 = food()    # creating the object (obj) of food() class
print(obj1.food_name)  # accessing the class variable


food.food_name = "Mutton" #class variable CAN be only change by class name

obj2 = food()
print(obj2.food_name) # class object (obj2) can access the class data
```

# Constructor function

```python
# EX-2 : Constructor function (__init__)

#DEF: Constructor function automatically get called/excuted whenever object is created
# Constructor function is used for intializing the variable, while creating the new object

#self is a variable which points to the current object

class Bike:
    No_tyre = 2    # class variable

    def __init__(self):
        print("Iam Constructor function")      #self points to the currect object whose defining the function

obj1 = Bike()
obj2 = Bike() # intitalize variable for the object & class variable value is same for the all objects
```

```python
# instance variable value is DIFFERENT for every object
# But class variable value is same for the all objects

class Bike:
    No_tyre = 2    # class variable

    def __init__(self, name):
        self.bike_name = name     # instance variable
        print(self.bike_name)      # printing instance variable

obj1 = Bike("Pulsur")   # Pulsur will store in instance variable and gets prints
obj2 = Bike("GT")   # intitalize instance variable for the object

# for every object instance variable is different then self will point to the object whoever calling the object
# internally self will be replaced by obj that is called
```

```python
class Bike:
    No_tyre = 2    # class variable

    def __init__(self, name, price):
        self.bike_name = name     # instance variable
        self.bike_price = price
        print(self.bike_name,self.bike_price)

obj1 = Bike("Pulsur",120000)
obj2 = Bike("GT",5000000)    
```

# Member function

```python
# EX-3: Member function  (display)

# Menmber function don't get called automatically, we need to call them ourself
# here also self is important & we use "display"

class Bike:
    No_tyre = 2    # class variable

    def __init__(self, name, price):
        self.bike_name = name     # instance variable
        self.bike_price = price
        print(self.bike_name,self.bike_price)

    def display(self):
        print("Iam a member function")

obj1 = Bike("Pulsur",120000)
obj2 = Bike("GT",5000000) 

obj1.display()
obj2.display()  # for member function we need to call 
```
