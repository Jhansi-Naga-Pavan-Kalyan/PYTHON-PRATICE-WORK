### tell() Method:

```python
# tell() tells us the current position of the file cursor.
with open("student.txt", "r") as file:
    print(file.tell())
    data = file.read(5)
    print(data)
    print(file.tell())
```

### seek() Method

```python
# seek() moves the file cursor to a particular position.
with open("student.txt", "r") as file:
    print(file.read(5))
    file.seek(0)
    print(file.read(5))
```

### tell() + seek() Together

```python
with open("student.txt", "r") as file:
    print("Position:", file.tell())
    print(file.read(3))
    print("Position:", file.tell())
    file.seek(0)
    print("Position:", file.tell())
    print(file.read(3))
```

### Check Whether File Exists

```python
# We can use the os module.
import os
if os.path.exists("student.txt"):
    print("File exists")
else:
    print("File does not exist")
```

### Delete a File

```python
# Be careful with this program because it permanently removes the file.
import os
if os.path.exists("student.txt"):
    os.remove("student.txt")
    print("File deleted")
else:
    print("File does not exist")
```

### Rename a File

### File Not Found Exception

```python
try:
    with open("abc.txt", "r") as file:
        print(file.read())
except FileNotFoundError:
    print("File not found")
```

### Handling Multiple Exceptions

```python
try:
    a = int(input("Enter number: "))
    with open("student.txt", "r") as file:
        print(file.read())
except ValueError:
    print("Please enter a valid number")
except FileNotFoundError:
    print("File not found")
```

```python
# Convert File Content to Uppercase
with open("student.txt", "w") as file:
    file.write("Ravi is a student.\n")
    file.write("Bhargavi is learning Python.\n")
    file.write("Python is easy to learn.")
# Read the file
with open("student.txt", "r") as file:
    data = file.read()
# Convert content to uppercase
data = data.upper()
# Write uppercase content back to the file
with open("student.txt", "w") as file:
    file.write(data)
# Display the result
print("File converted to uppercase successfully!")
print("\nFile Content:")
print(data)
```

```python
# Convert File Content to Lowercase
with open("student.txt", "w") as file:
    file.write("RAVI IS A STUDENT.\n")
    file.write("BHARGAVI IS LEARNING PYTHON.\n")
    file.write("PYTHON IS EASY TO LEARN.")
# Read the file
with open("student.txt", "r") as file:
    data = file.read()
# Convert content to lowercase
data = data.lower()
# Write the lowercase content back to the file
with open("student.txt", "w") as file:
    file.write(data)
# Display the result
print("File converted to lowercase successfully!")
print("\nFile Content:")
print(data)
```
