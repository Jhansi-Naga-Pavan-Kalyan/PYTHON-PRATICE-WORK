# polymorphism

```python
# one function can perform many operations
# the word polymorphism means "many form"
# in polymorphism, same function or same oprator can perform many different task.

#ex: we  use "+" for different task (opending, adding)

# Type of polymorphism:
# 1. function overloading
# 2. Operator overloading
# 3. Method overriding
```

## 1. function overloading

## using default Arguments for function overloading

```python
# var2 is default arguments
def fun(var1, var2 = None):
    if var2 is not None:
        print(var1, var2)
    else:
        print(var1)

fun(10)
fun(10,20) # same function is perfoeming different tasks
```

```python
def fun(var1, var2 = None, var3= None):
    if var2 is not None:
        print(var1, var2, var3)
    else:
        print(var1)

fun(10)
fun(10,20)
fun(10,20,30)
```

```python
# Exam-2
def calculator(a = 0, b = 0, c =0, d = 0):  # default argument
    return a + b + c + d

print(calculator())
print(calculator(100))
print(calculator(100,200))
print(calculator(100,200,300))
print(calculator(100,200,300,400))
```

## *arg for function overloading

```python
# *arg can also use as function overloading
def calculator(*arg):
    return sum(arg)

print(calculator())
print(calculator(100))
print(calculator(100,200))
print(calculator(100,200,300))
print(calculator(100,200,300,400))
print(calculator(100,200,300,400,500,600,700,800,900))
```

## function overloading using decorator

```python
#function overloading using decorator

from multipledispatch import dispatch  #dispatch is a decorator, multipledispatch is library

@dispatch(int)
def fun(var1):
    print(var1)

@dispatch(int, int)
def fun(var1, var2):
    print(var1, var2)

@dispatch(float, float, int)
def fun(f1, f2, i1):
    print(f1, f2, i1)

fun(10)
fun(10,20)
fun(3.33,33.3,333)
```

# 2. Operator overloading

```python
# we will perform different operations on the user defined objects using operators.
```

```python
class Dosa:
    def __init__(self, variety, price):
        self.Variety = variety
        self.Price = price
    def __add__(self, d2_val):   # due to this the program got successful
        total_price = self.Price + d2_val.Price
        return total_price
    def __sub__(self, d2_val):   # python as already defined this function
        total_price = self.Price - d2_val.Price
        return total_price
    def __mul__(self, d2_val):   # python as already defined this function
        total_price = self.Price * d2_val.Price
        return total_price
    def __truediv__(self, d2_val):   # python as already defined this function
        total_price = self.Price / d2_val.Price
        return total_price
    def __floordiv__(self, d2_val):   # python as already defined this function
        total_price = self.Price // d2_val.Price
        return total_price
    def __mod__(self, d2_val):   # python as already defined this function
        total_price = self.Price % d2_val.Price
        return total_price

d1 = Dosa("Masala", 60)
d2 = Dosa("IceCream", 110)

print(d1+d2)
print(d2 - d1)
print(d1 * d2)
print(d2 / d1) # __truediv__
print(d2 // d1) # __floordiv__
print(d1 % d2)  # __mod__
#print(d1 ** d2) # __pow__
```

```python
# comparison Operators overloading

class Dosa:
    def __init__(self, variety, price):
        self.Variety = variety
        self.Price = price
    def __gt__(self, d2_val):   
        total_price = self.Price > d2_val.Price
        return total_price
    def __lt__(self, d2_val):   
        total_price = self.Price < d2_val.Price
        return total_price
    def __eq__(self, d2_val):   
        total_price = self.Price == d2_val.Price
        return total_price
    def __ge__(self, d2_val):   
        total_price = self.Price >= d2_val.Price
        return total_price
    def __le__(self, d2_val):   
        total_price = self.Price <= d2_val.Price
        return total_price
    


d1 = Dosa("Masala", 60)
d2 = Dosa("IceCream", 110)

print(d1 > d2) # __gt__
print(d1 < d2) # __lt__
print(d1 == d2) # __eq__
print(d1 >= d2) # __ge__
print(d1 <= d2) # __le__
```

# 3. Method Overriding

```python
#Method overriding occurs when a child class defines a method with the same name as a method in its parent class. 
```

```python
class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    def sound(self):
        print("Dog barks")

d = Dog()
d.sound()
```
