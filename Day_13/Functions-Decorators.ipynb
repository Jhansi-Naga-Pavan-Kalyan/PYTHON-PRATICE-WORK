# Recurise function

```python
# DEF: A function calling itself directly or indirectly
# termination condition is must in recursion
# in recurive call we must chnage the parameter value
```

```python
# ex-1
def foo(i):
    if i < 1:
        return
    print(i)
    foo(i-1)    # recurise call
foo(6)
```

```python
# ex-2
def foo(i):
    if i < 1:  
        return
    foo(i-1)
    print(i)
foo(6)
```

```python
# ex-2
def foo(i):
    if i < 1:  # termination condition r base condition also must be there in recusion fun
        return
    print(i)
    foo(i-1)     # when ur calling recuserive fun u have to increment r decrement
    print(i)
foo(6)
```

# Enumerate Function

```python
# enumerate() function adds the counter (index) with the list/tuple items.
```

```python
fruits_names = ["MamidiPandu","PanasaPandu","JamaPandu"]

for name in fruits_names:
    print(name)
```

```python
fruits_names = ["MamidiPandu","PanasaPandu","JamaPandu"]

for index, name in enumerate(fruits_names):    # by using enumerate we can index value 
    print(index, name)
```

```python
fruits_names = ["MamidiPandu","PanasaPandu","JamaPandu"]

for index, name in enumerate(fruits_names, start = 101):  # whtever numbing u want to start u can pass by using start
    print(index, name)
```

# zip() function

```python
# zip() function is used to combine/zip mulltiple list/tuple items, element-wise
```

```python
Hero = ["Mahesh", "Malikarjun", "Pavan Kalyan", "Balajji"]
Heronies = ["Anushka", "Samanta", "Rashmika", "kajal"]


pairs = zip(Hero, Heronies)
print(list(pairs))
```

```python
Hero = ["Mahesh", "Malikarjun", "Pavan Kalyan", "Balajji"]
Heronies = ["Anushka", "Samanta", "Rashmika", "kajal"]
Movie = ["MA","MS","PR","BK"]

pairs = zip(Hero, Heronies, Movie)  # we can zip together using differnt lists make sure number of elements must be same 
print(list(pairs))
```

# Decorators

```python
def decorator_fun(func):   # func => cake()
    def wrapper_fun():   # wrapper function add the extra functionality
        func()
        print("add cherry")    # Extra functionality
        print("add chocolate")
    return wrapper_fun          # modified function

def cake():
    print("take the flour")
    print("add eggs and other ingredient")
    print("put the cake in oven")

decorated_function = decorator_fun(cake)
decorated_function()
```

```python
def decorator_fun(func):   # func => cake()
    def wrapper_fun():# wrapper function add the extra functionality
        print("Buy ingrdents")   # we can add extra functionality before nd after the original fun
        func()
        print("add cherry")    # Extra functionality
        print("add chocolate")
    return wrapper_fun          # modified function

def cake():
    print("take the flour")
    print("add eggs and other ingredient")
    print("put the cake in oven")

decorated_function = decorator_fun(cake)
decorated_function()
```

```python
# we repersent decorator with the help of @

# ex-2: use of @ to repersent the decorator

def decorator_fun(func):   # func => cake()
    def wrapper_fun():   # wrapper function add the extra functionality
        print("Buy ingrdents")
        func()
        print("add cherry")    # Extra functionality
        print("add chocolate")
    return wrapper_fun 

@decorator_fun        # if we use '@' with decorator function then it gets decorated
def cake():
    print("take the flour")
    print("add eggs and other ingredient")
    print("put the cake in oven")

cake()
```

```python
# multiple decorator for same function

def decorator_fun1(func):   # func => cake()
    def wrapper_fun():   # wrapper function add the extra functionality
        print("Buy ingrdents")
        func()
    return wrapper_fun

def decorator_fun2(func):   # func => cake()
    def wrapper_fun():   # wrapper function add the extra functionality
        func()
        print("add cherry")    # Extra functionality
        print("add chocolate")
    return wrapper_fun

@decorator_fun1
@decorator_fun2

def cake():
    print("take the flour")
    print("add eggs and other ingredient")
    print("put the cake in oven")

cake()
```
