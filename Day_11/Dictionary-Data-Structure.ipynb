# Dictionary data structure

### Different ways to create dictionary

```python
empty_dict = {}
empty_dict2 = dict()

print(empty_dict)
print(type(empty_dict2))
```

```python
dupl_values_dict = {1: 'mamidipandu', 2 : 'mamidipandu', 3: 'Batai'}  # duplicate values are allowed
print(dupl_values_dict)
dupl_keys_dict = {1: 'mamidipandu', 1 : 'Arratipandu', 3: 'Batai'}  # duplicate keys are not allowed
print(dupl_keys_dict)
```

```python
int_dict = {1: 'mamidipandu', 2 : 'Aratipandu', 3: 'Batai'}  # keys are defined as int
print(int_dict)

float_dict = {1.2 : 'mamidipandu', 2.3 : 'Aratipandu', 3.4 : 'Batai'} # keys are defined as float
print(float_dict)

string_dict = {'mango' : 'mamidipandu', 'banana' : 'Aratipandu', 'orange' : 'Batai'} # keys are defined as string
print(string_dict)

hetro_dict = {1 : 'mamidipandu', 'banana' : 'Aratipandu', 3.4 : 'Batai'} # keys are defined as hetro
print(hetro_dict)

tuple_dict = {(10,20) : 'mamidipandu', 'banana' : 'Aratipandu', 3.4 : 'Batai'} # keys are defined as tuple
print(tuple_dict)

# tuple can be used as the key when u use multiple values as a key we  use tuple
#list & set cannot be used as the key
#keys must be immutable due to hash table

# values can be mutable and immutable

dict_comp = {x : x**2 for x in range(1,6)}
print(dict_comp)

int_dict[3]
```

```python
#coverting to dictionary

list_of_tuple = [('Gopichand',6),('chandar',90),('Dinesh',100)]

dict1 = dict(list_of_tuple)
print(dict1)

list_of_list = [['Gopichand',6],['chandar',90],['Dinesh',100]]
dict2 = dict(list_of_list)
print(dict2)

tuple_of_tuple = (('Gopichand',6),('chandar',90),('Dinesh',100))
dict3 = dict(tuple_of_tuple)
print(dict3)

tuple_of_list = (['Gopichand',6],['chandar',90],['Dinesh',100])
dict4 = dict(tuple_of_list)
print(dict4)

set_of_tuple = {('Gopichand',6),('chandar',90),('Dinesh',100)}
dict5 = dict(set_of_tuple)
print(dict5)
```

# Accessing dictionary items 

```python
drink_items = {
                "Thumpsup" : 50,
                "Sugarcane" : 20,
                 "coconutwater" : 60,
                 "water" : 20,
                  "rosemilk" : 30
}

val = drink_items["water"]  # if value is not their it will show error
print(val)

val2 = drink_items.get("rosemilk")
print(val2)

val2 = drink_items.get("coke","key not avaliable") # if key is not their it will not show error
print(val2)
```

```python
#to access the keys
for i in drink_items:
    print(i)
```

```python
#to access the keys and values
for i in drink_items:
    print(i, drink_items[i])
```

```python
# to access all the keys we use key() functions
print(drink_items.keys())
```

```python
# to access all the values we use values() functions
print(drink_items.values())
```

```python
# to access all the items we use items() functions
print(drink_items.items())
```

```python
# nested dictionary
nested_dict = {
                100 : {"Hello":10, "World" : 20},
                200 : {"A" : 50 , "B" : 60}
}

nested_dict[100]
nested_dict[100]["World"]
```

# insert & update opeartions on dictionary

```python
x_dict = {"name" : "Chinki", "age" : 80}

x_dict["age"] = 180  # if key is avaliable it will update
print(x_dict)

x_dict["city"] = "Kurnool"  # if key is not avaliable it will insert
print(x_dict)
```

```python
# insert & update using update() function
# updat() can insert & update multiple items

x_dict.update({'age' : 18, 'occupation' : 'IAS', "name" : "minki", "Salary" : 1.5})
print(x_dict)
```

# delete the items from dictionary

```python
# pop(), popitem(), clear(), del()

x_dict
```

```python
# pop() removes the items whhose key is specified
x_dict.pop("Salary")
print(x_dict)
```

```python
# popitem() removes last  element
x_dict.popitem()
print(x_dict)
```

```python
# clear() clears all the elemments
x_dict.clear()
print(x_dict)
```

```python
# deletes whole dictionary
del x_dict
```
