# Python Exception Handling

```python
# An exception is a problem that happens while the program is running.
a = 10
b = 0
print(a / b) # Normally, the program stops when an exception occurs.
#Exception handling allows us to handle this problem without suddenly stopping the program.
```

### try and except

```python
try:
    a = 10
    b = 0
    result = a / b
    print(result)
except:
    print("Something went wrong")
```

```python
# It is better to specify the exact exception.
try:
    a = 10
    b = 0
    result = a / b
    print(result)
except ZeroDivisionError:
    print("Cannot divide by zero")
```

### ValueError

```python
# A ValueError occurs when the value is not appropriate for an operation.
try:
    age = int("hello")
    print(age)
except ValueError:
    print("Please enter a valid number")
```

### Handling User Input

```python
try:
    age = int(input("Enter your age: "))
    print("Your age is:", age)
except ValueError:
    print("Please enter numbers only")
```

### Multiple except Blocks

```python
# We can handle different exceptions separately.
try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    result = a / b
    print("Result:", result)
except ValueError:
    print("Please enter numbers only")
except ZeroDivisionError:
    print("Cannot divide by zero")
```

### else

```python
# The else block runs only when there is no exception.
try:
    a = 10
    b = 2
    result = a / b
except ZeroDivisionError:
    print("Cannot divide by zero")
else:
    print("Result:", result)
```

### finally

```python
# The finally block always executes, whether an error occurs or not.
try:
    a = 10
    b = 0
    result = a / b
    print(result)
except ZeroDivisionError:
    print("Cannot divide by zero")
finally:
    print("Program completed")
```

### try + except + else + finally

```python
try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    result = a / b
except ValueError:
    print("Please enter numbers only")
except ZeroDivisionError:
    print("Cannot divide by zero")
else:
    print("Result:", result)
finally:
    print("Execution completed")
```

### Getting the Error Message using as e

```python
# We can store the exception in a variable.
try:
    result = 10 / 0
except ZeroDivisionError as e:
    print("Error:", e)
```

### Multiple Exceptions in One except

```python
# If you want the same message for multiple exceptions:
try:
    a = int(input("Enter number: "))
    print(10 / a)
except (ValueError, ZeroDivisionError):
    print("Invalid input")
```

### raise Keyword

```python
#Sometimes we want to manually create an exception.
#We use raise.
age = -5
if age < 0:
    raise ValueError("Age cannot be negative")
print("Age:", age)
```

### raise with a Function

```python
def check_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative")
    print("Valid age:", age)
try:
    check_age(-5)
except ValueError as e:
    print("Error:", e)
```

### Bank Account Example

```python
def withdraw(balance, amount):
    if amount > balance:
        raise ValueError("Insufficient balance")
    return balance - amount
try:
    balance = 5000
    amount = 7000
    balance = withdraw(balance, amount)
    print("Remaining balance:", balance)
except ValueError as e:
    print("Transaction failed:", e)
```

### IndexError Example

```python
numbers = [10, 20, 30]
try:
    print(numbers[5])
except IndexError:
    print("Index does not exist")
```

### Example of KeyError

```python
student = {
    "name": "Ravi",
    "age": 20
}
try:
    print(student["marks"])
except KeyError:
    print("Marks key does not exist")
```

### Example of TypeError

```python
try:
    result = 10 + "20"
    print(result)
except TypeError:
    print("Cannot add integer and string")
```
